```python
# ============================================================
# APEXPLANET DATA ANALYTICS INTERNSHIP
# TASK 1 - SUPERSTORE SALES
# DATA CLEANING + STATISTICAL ANALYSIS + EDA
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 70)
print("SUPERSTORE SALES - EXPLORATORY DATA ANALYSIS")
print("=" * 70)

print("\nLibraries imported successfully!")


# ============================================================
# 1. LOAD CLEANED DATASET
# ============================================================

DATA_PATH = "../data/processed/superstore_cleaned.csv"

print("\nLoading cleaned dataset...")
print("File:", DATA_PATH)

df = pd.read_csv(DATA_PATH, encoding="latin-1")

print("\nDataset loaded successfully!")


# ============================================================
# 2. BASIC DATASET INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("DATASET OVERVIEW")
print("=" * 70)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())


# ============================================================
# 3. DATASET INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

df.info()


# ============================================================
# 4. STATISTICAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("STATISTICAL SUMMARY")
print("=" * 70)

display(df.describe().T)


# ============================================================
# 5. CATEGORICAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CATEGORICAL SUMMARY")
print("=" * 70)

categorical_columns = df.select_dtypes(
    include="object"
).columns

display(df[categorical_columns].describe().T)


# ============================================================
# 6. MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

missing_values = df.isnull().sum()

display(
    missing_values.to_frame(
        "Missing Values"
    )
)

print(
    "\nTotal missing values:",
    missing_values.sum()
)


# ============================================================
# 7. DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("DUPLICATE ANALYSIS")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print(
    "Duplicate rows:",
    duplicate_count
)


# ============================================================
# 8. DATE CONVERSION
# ============================================================

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce"
)

print("\nOrder Date and Ship Date converted successfully.")


# ============================================================
# 9. CREATE TIME FEATURES
# ============================================================

df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month_Name"] = df["Order Date"].dt.strftime("%b")
df["Quarter"] = df["Order Date"].dt.quarter

print("Time features created successfully.")


# ============================================================
# 10. CREATE PROFIT MARGIN
# ============================================================

df["Profit_Margin"] = np.where(
    df["Sales"] != 0,
    (df["Profit"] / df["Sales"]) * 100,
    0
)

print("Profit margin calculated successfully.")


# ============================================================
# 11. BUSINESS KPIs
# ============================================================

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()

average_sales = df["Sales"].mean()
average_profit = df["Profit"].mean()

profit_margin = (
    total_profit / total_sales
) * 100


print("\n" + "=" * 70)
print("BUSINESS KPIs")
print("=" * 70)

print(
    f"Total Sales       : ${total_sales:,.2f}"
)

print(
    f"Total Profit      : ${total_profit:,.2f}"
)

print(
    f"Total Quantity    : {total_quantity:,.0f}"
)

print(
    f"Average Sales     : ${average_sales:,.2f}"
)

print(
    f"Average Profit    : ${average_profit:,.2f}"
)

print(
    f"Profit Margin     : {profit_margin:.2f}%"
)


# ============================================================
# 12. HISTOGRAM - SALES
# ============================================================

plt.figure(figsize=(10, 6))

plt.hist(
    df["Sales"],
    bins=30,
    edgecolor="black"
)

plt.title(
    "Distribution of Sales",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


# ============================================================
# 13. HISTOGRAM - PROFIT
# ============================================================

plt.figure(figsize=(10, 6))

plt.hist(
    df["Profit"],
    bins=30,
    edgecolor="black"
)

plt.title(
    "Distribution of Profit",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


# ============================================================
# 14. HISTOGRAM - QUANTITY
# ============================================================

plt.figure(figsize=(10, 6))

plt.hist(
    df["Quantity"],
    bins=10,
    edgecolor="black"
)

plt.title(
    "Distribution of Quantity",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Quantity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


# ============================================================
# 15. BOXPLOT - SALES
# ============================================================

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["Sales"]
)

plt.title(
    "Sales Distribution and Outliers",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Sales")

plt.tight_layout()
plt.show()


# ============================================================
# 16. BOXPLOT - PROFIT
# ============================================================

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["Profit"]
)

plt.title(
    "Profit Distribution and Outliers",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Profit")

plt.tight_layout()
plt.show()


# ============================================================
# 17. CATEGORY ANALYSIS
# ============================================================

category_summary = df.groupby(
    "Category"
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Average_Discount=("Discount", "mean")
)

print("\n" + "=" * 70)
print("CATEGORY ANALYSIS")
print("=" * 70)

display(category_summary)


# ============================================================
# 18. SALES BY CATEGORY
# ============================================================

category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

category_sales.plot(
    kind="bar",
    edgecolor="black"
)

plt.title(
    "Sales by Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 19. PROFIT BY CATEGORY
# ============================================================

category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

category_profit.plot(
    kind="bar",
    edgecolor="black"
)

plt.title(
    "Profit by Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 20. REGION ANALYSIS
# ============================================================

region_summary = df.groupby(
    "Region"
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Average_Discount=("Discount", "mean")
)

print("\n" + "=" * 70)
print("REGION ANALYSIS")
print("=" * 70)

display(region_summary)


# ============================================================
# 21. SALES BY REGION
# ============================================================

region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

region_sales.plot(
    kind="bar",
    edgecolor="black"
)

plt.title(
    "Sales by Region",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 22. PROFIT BY REGION
# ============================================================

region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

region_profit.plot(
    kind="bar",
    edgecolor="black"
)

plt.title(
    "Profit by Region",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Region")
plt.ylabel("Total Profit")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 23. CUSTOMER SEGMENT ANALYSIS
# ============================================================

segment_summary = df.groupby(
    "Segment"
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
)

print("\n" + "=" * 70)
print("CUSTOMER SEGMENT ANALYSIS")
print("=" * 70)

display(segment_summary)


# ============================================================
# 24. SALES BY CUSTOMER SEGMENT
# ============================================================

segment_sales = (
    df.groupby("Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

segment_sales.plot(
    kind="bar",
    edgecolor="black"
)

plt.title(
    "Sales by Customer Segment",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 25. SHIPPING MODE ANALYSIS
# ============================================================

ship_mode_summary = df.groupby(
    "Ship Mode"
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Order_Count=("Order ID", "nunique")
)

print("\n" + "=" * 70)
print("SHIPPING MODE ANALYSIS")
print("=" * 70)

display(ship_mode_summary)


# ============================================================
# 26. SALES BY SHIPPING MODE
# ============================================================

ship_mode_sales = (
    df.groupby("Ship Mode")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

ship_mode_sales.plot(
    kind="bar",
    edgecolor="black"
)

plt.title(
    "Sales by Shipping Mode",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Shipping Mode")
plt.ylabel("Total Sales")

plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ============================================================
# 27. SUB-CATEGORY ANALYSIS
# ============================================================

subcategory_summary = (
    df.groupby("Sub-Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values(
        "Total_Sales",
        ascending=False
    )
)

print("\n" + "=" * 70)
print("SUB-CATEGORY ANALYSIS")
print("=" * 70)

display(subcategory_summary)


# ============================================================
# 28. TOP 10 SUB-CATEGORIES
# ============================================================

top_subcategories = (
    subcategory_summary
    .head(10)
)

plt.figure(figsize=(12, 7))

top_subcategories[
    "Total_Sales"
].sort_values().plot(
    kind="barh"
)

plt.title(
    "Top Sub-Categories by Sales",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Sales")
plt.ylabel("Sub-Category")

plt.tight_layout()
plt.show()


# ============================================================
# 29. CORRELATION MATRIX
# ============================================================

numeric_columns = df[
    [
        "Sales",
        "Quantity",
        "Discount",
        "Profit"
    ]
]

correlation = numeric_columns.corr()

print("\n" + "=" * 70)
print("CORRELATION MATRIX")
print("=" * 70)

display(correlation)


# ============================================================
# 30. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title(
    "Correlation Matrix Heatmap",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()


# ============================================================
# 31. MONTHLY SALES TREND
# ============================================================

monthly_sales = (
    df.groupby(
        df["Order Date"].dt.to_period("M")
    )["Sales"]
    .sum()
)

monthly_sales.index = (
    monthly_sales.index.to_timestamp()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title(
    "Monthly Sales Trend",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Date")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# 32. MONTHLY PROFIT TREND
# ============================================================

monthly_profit = (
    df.groupby(
        df["Order Date"].dt.to_period("M")
    )["Profit"]
    .sum()
)

monthly_profit.index = (
    monthly_profit.index.to_timestamp()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_profit.index,
    monthly_profit.values,
    marker="o"
)

plt.title(
    "Monthly Profit Trend",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Date")
plt.ylabel("Profit")

plt.xticks(rotation=45)

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# 33. YEARLY SALES
# ============================================================

yearly_sales = (
    df.groupby("Year")["Sales"]
    .sum()
)

print("\n" + "=" * 70)
print("YEARLY SALES")
print("=" * 70)

display(
    yearly_sales.to_frame("Sales")
)

plt.figure(figsize=(9, 6))

yearly_sales.plot(
    kind="bar",
    edgecolor="black"
)

plt.title(
    "Yearly Sales",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Year")
plt.ylabel("Sales")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 34. YEARLY PROFIT
# ============================================================

yearly_profit = (
    df.groupby("Year")["Profit"]
    .sum()
)

print("\n" + "=" * 70)
print("YEARLY PROFIT")
print("=" * 70)

display(
    yearly_profit.to_frame("Profit")
)

plt.figure(figsize=(9, 6))

yearly_profit.plot(
    kind="bar",
    edgecolor="black"
)

plt.title(
    "Yearly Profit",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Year")
plt.ylabel("Profit")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 35. TOP 10 PRODUCTS BY SALES
# ============================================================

top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\n" + "=" * 70)
print("TOP 10 PRODUCTS BY SALES")
print("=" * 70)

display(
    top_products.to_frame("Sales")
)

plt.figure(figsize=(12, 7))

top_products.sort_values().plot(
    kind="barh"
)

plt.title(
    "Top 10 Products by Sales",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Sales")
plt.ylabel("Product")

plt.tight_layout()
plt.show()


# ============================================================
# 36. BOTTOM 10 PRODUCTS BY PROFIT
# ============================================================

bottom_products = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values()
    .head(10)
)

print("\n" + "=" * 70)
print("BOTTOM 10 PRODUCTS BY PROFIT")
print("=" * 70)

display(
    bottom_products.to_frame("Profit")
)

plt.figure(figsize=(12, 7))

bottom_products.plot(
    kind="barh"
)

plt.title(
    "Bottom 10 Products by Profit",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Profit")
plt.ylabel("Product")

plt.tight_layout()
plt.show()


# ============================================================
# 37. LOSS-MAKING ORDERS
# ============================================================

loss_orders = df[
    df["Profit"] < 0
]

number_of_loss_orders = len(loss_orders)

total_loss = loss_orders["Profit"].sum()

print("\n" + "=" * 70)
print("LOSS-MAKING ANALYSIS")
print("=" * 70)

print(
    "Number of loss-making records:",
    number_of_loss_orders
)

print(
    f"Total loss: ${total_loss:,.2f}"
)


# ============================================================
# 38. LOSS BY CATEGORY
# ============================================================

loss_by_category = (
    loss_orders
    .groupby("Category")["Profit"]
    .sum()
    .sort_values()
)

print("\nLoss by Category:")

display(
    loss_by_category.to_frame("Loss")
)


# ============================================================
# 39. DISCOUNT VS PROFIT
# ============================================================

discount_profit = (
    df.groupby("Discount")["Profit"]
    .mean()
    .sort_index()
)

print("\n" + "=" * 70)
print("DISCOUNT VS AVERAGE PROFIT")
print("=" * 70)

display(
    discount_profit.to_frame(
        "Average Profit"
    )
)

plt.figure(figsize=(10, 6))

plt.plot(
    discount_profit.index,
    discount_profit.values,
    marker="o"
)

plt.title(
    "Average Profit by Discount",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Discount")
plt.ylabel("Average Profit")

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# 40. CATEGORY PROFIT MARGIN
# ============================================================

category_margin = (
    df.groupby("Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
)

category_margin["Profit_Margin"] = (
    category_margin["Profit"] /
    category_margin["Sales"]
) * 100

print("\n" + "=" * 70)
print("CATEGORY PROFIT MARGIN")
print("=" * 70)

display(category_margin)


# ============================================================
# 41. REGION PROFIT MARGIN
# ============================================================

region_margin = (
    df.groupby("Region")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
)

region_margin["Profit_Margin"] = (
    region_margin["Profit"] /
    region_margin["Sales"]
) * 100

print("\n" + "=" * 70)
print("REGION PROFIT MARGIN")
print("=" * 70)

display(region_margin)


# ============================================================
# 42. AUTOMATIC KEY INSIGHTS
# ============================================================

best_category_sales = category_sales.idxmax()
best_category_sales_value = category_sales.max()

best_category_profit = category_profit.idxmax()
best_category_profit_value = category_profit.max()

best_region_sales = region_sales.idxmax()
best_region_sales_value = region_sales.max()

best_region_profit = region_profit.idxmax()
best_region_profit_value = region_profit.max()

best_segment = segment_sales.idxmax()
best_segment_value = segment_sales.max()

best_product = top_products.idxmax()
best_product_sales = top_products.max()

worst_product = bottom_products.idxmin()
worst_product_profit = bottom_products.min()

highest_sales_month = monthly_sales.idxmax()
highest_sales_month_value = monthly_sales.max()

highest_profit_month = monthly_profit.idxmax()
highest_profit_month_value = monthly_profit.max()

sales_profit_corr = correlation.loc[
    "Sales",
    "Profit"
]

discount_profit_corr = correlation.loc[
    "Discount",
    "Profit"
]


print("\n" + "=" * 70)
print("5 KEY BUSINESS INSIGHTS")
print("=" * 70)

print(
    f"""
1. CATEGORY PERFORMANCE
The {best_category_sales} category generated the highest
total sales of ${best_category_sales_value:,.2f}.

2. REGIONAL PERFORMANCE
The {best_region_sales} region generated the highest
total sales of ${best_region_sales_value:,.2f}.

3. CUSTOMER SEGMENT
The {best_segment} segment generated the highest
sales of ${best_segment_value:,.2f}.

4. PRODUCT PERFORMANCE
The top product by sales was "{best_product}",
generating ${best_product_sales:,.2f} in sales.

5. PROFITABILITY
The overall profit margin was {profit_margin:.2f}%.
The correlation between Sales and Profit was
{sales_profit_corr:.2f}.
"""
)


# ============================================================
# 43. ADDITIONAL INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("ADDITIONAL BUSINESS INSIGHTS")
print("=" * 70)

print(
    f"\nHighest sales month: "
    f"{highest_sales_month.strftime('%B %Y')} "
    f"(${highest_sales_month_value:,.2f})"
)

print(
    f"Highest profit month: "
    f"{highest_profit_month.strftime('%B %Y')} "
    f"(${highest_profit_month_value:,.2f})"
)

print(
    f"\nLowest-profit product: "
    f"{worst_product}"
    f" (${worst_product_profit:,.2f})"
)

print(
    f"\nDiscount-Profit correlation: "
    f"{discount_profit_corr:.2f}"
)

print(
    f"\nTotal loss from loss-making records: "
    f"${total_loss:,.2f}"
)


# ============================================================
# 44. SAVE EDA-ENRICHED DATASET
# ============================================================

OUTPUT_PATH = "../data/processed/superstore_eda_final.csv"

df.to_csv(
    OUTPUT_PATH,
    index=False
)

print("\n" + "=" * 70)
print("EDA DATASET SAVED")
print("=" * 70)

print(
    f"\nSaved to:\n{OUTPUT_PATH}"
)


# ============================================================
# 45. FINAL MESSAGE
# ============================================================

print("\n" + "=" * 70)
print("DAY 3-6 EDA COMPLETED SUCCESSFULLY!")
print("=" * 70)

print(
    """
✓ Cleaned dataset loaded
✓ Statistical summary generated
✓ Dataset information checked
✓ Missing values checked
✓ Duplicate values checked
✓ Date analysis completed
✓ Time features created
✓ Histograms created
✓ Boxplots created
✓ Category analysis completed
✓ Region analysis completed
✓ Customer segment analysis completed
✓ Shipping mode analysis completed
✓ Sub-category analysis completed
✓ Correlation matrix created
✓ Correlation heatmap created
✓ Monthly sales trend created
✓ Monthly profit trend created
✓ Yearly analysis completed
✓ Product analysis completed
✓ Loss analysis completed
✓ Discount analysis completed
✓ Profit margin calculated
✓ 5 key insights generated
✓ Final EDA dataset saved

SUPERSTORE EDA COMPLETE!
"""
)
```
